```python

# Emergency Dispatch Record Search

# -----------------------------------------
# 1. STUDENT-SPECIFIC INPUTS
# -----------------------------------------

last_name = input("Enter your last name: ").strip().upper()
student_id = input("Enter your student ID number: ").strip()
favorite_artist = input("Enter your favorite artist: ").strip().upper()

# Check if the student ID contains only numbers
if not student_id.isdigit() or student_id == "":
    print("Invalid student ID. Please enter numbers only.")
else:
    # -----------------------------------------
    # 2. GENERATE STUDENT-SPECIFIC VALUES
    # -----------------------------------------

    seed_digit = int(student_id[-1])
    id_sum = sum(int(digit) for digit in student_id)
    name_length = len(last_name)

    artist_length = len(favorite_artist)

    # Get the alphabetical value of each letter
    artist_sum = 0

    for letter in favorite_artist:
        if letter.isalpha() and letter.isascii():
            artist_sum += ord(letter) - ord('A') + 1

    print("\n===== STUDENT INFORMATION =====")
    print("Last Name:", last_name)
    print("Student ID:", student_id)
    print("Favorite Artist:", favorite_artist)
    print("SEED_DIGIT:", seed_digit)
    print("ID_SUM:", id_sum)
    print("NAME_LENGTH:", name_length)
    print("ARTIST_LENGTH:", artist_length)
    print("ARTIST_SUM:", artist_sum)

    # -----------------------------------------
    # 3. GENERATE INCIDENT RECORDS
    # -----------------------------------------

    number_of_records = 15

    # Use student information to create a starting number
    start_number = (id_sum * name_length + artist_sum) * 10

    original_records = []

    for i in range(number_of_records):
        incident_number = start_number + i + 1
        original_records.append(incident_number)

    print("\n===== ORIGINAL DATASET =====")
    print(original_records)

    # -----------------------------------------
    # 4. ROTATE THE INCIDENT RECORDS
    # -----------------------------------------

    rotation_point = seed_digit % number_of_records

    rotated_records = (
        original_records[rotation_point:]
        + original_records[:rotation_point]
    )

    print("\nRotation Point:", rotation_point)
    print("===== ROTATED DATASET =====")
    print(rotated_records)

    # -----------------------------------------
    # 5. GENERATE THE TARGET INCIDENT
    # -----------------------------------------

    target_index = (id_sum + artist_sum) % number_of_records
    target_incident = original_records[target_index]

    print("\nTarget Incident:", target_incident)

    # -----------------------------------------
    # 6. SEARCH THE ROTATED DATASET
    #    USING MODIFIED BINARY SEARCH
    # -----------------------------------------

    def rotated_binary_search(records, target):
        left = 0
        right = len(records) - 1
        comparisons = 0

        while left <= right:
            middle = (left + right) // 2

            comparisons += 1
            if records[middle] == target:
                return middle, comparisons

            comparisons += 1
            if records[left] <= records[middle]:

                if records[left] <= target < records[middle]:
                    right = middle - 1
                else:
                    left = middle + 1

            else:
                if records[middle] < target <= records[right]:
                    left = middle + 1
                else:
                    right = middle - 1

        return -1, comparisons

    position, binary_comparisons = rotated_binary_search(
        rotated_records, target_incident
    )

    print("\n===== BINARY SEARCH RESULT =====")

    if position != -1:
        print("Target incident found!")
        print("Position in rotated dataset:", position + 1)
    else:
        print("Target incident was not found.")

    print("Binary search comparisons:", binary_comparisons)

    # -----------------------------------------
    # 7. COMPARE WITH LINEAR SEARCH
    # -----------------------------------------

    def linear_search(records, target):
        comparisons = 0

        for i in range(len(records)):
            comparisons += 1

            if records[i] == target:
                return i, comparisons

        return -1, comparisons

    linear_position, linear_comparisons = linear_search(
        original_records, target_incident
    )

    print("\n===== LINEAR SEARCH RESULT =====")

    if linear_position != -1:
        print("Target incident found!")
        print("Position in original dataset:", linear_position + 1)
    else:
        print("Target incident was not found.")

    print("Linear search comparisons:", linear_comparisons)

    # -----------------------------------------
    # 8. FINAL COMPARISON
    # -----------------------------------------

    print("\n===== SEARCH COMPARISON =====")
    print("Binary Search Comparisons:", binary_comparisons)
    print("Linear Search Comparisons:", linear_comparisons)

    if binary_comparisons < linear_comparisons:
        print("Binary search used fewer comparisons.")
    elif binary_comparisons > linear_comparisons:
        print("Linear search used fewer comparisons.")
    else:
        print("Both searches used the same number of comparisons.")

    # -----------------------------------------
    # 9. TIME AND SPACE COMPLEXITY
    # -----------------------------------------

    print("\n===== TIME AND SPACE COMPLEXITY =====")
    print("Modified Binary Search:")
    print("  Average/Worst Time: O(log n)")
    print("  Auxiliary Space: O(1)")

    print("\nLinear Search:")
    print("  Average/Worst Time: O(n)")
    print("  Auxiliary Space: O(1)")

    print("\nDataset Generation and Rotation:")
    print("  Time Complexity: O(n)")
    print("  Additional Space: O(n)")
```

```python
def calculate_student_inputs(last_name, student_id, favorite_artist):
    """Calculates student-specific values based on input rules."""
    # Convert inputs to standard formats
    last_name = last_name.strip()
    student_id = str(student_id).strip()
    favorite_artist = favorite_artist.strip()
    
    # 1. SEED_DIGIT: Last digit of STUDENT_ID
    seed_digit = int(student_id[-1]) if student_id else 0
    
    # 2. ID_SUM: Sum of STUDENT_ID digits
    id_sum = sum(int(digit) for digit in student_id if digit.isdigit())
    
    # 3. NAME_LENGTH: Length of LAST_NAME
    name_length = len(last_name)
    
    # 4. ARTIST_LENGTH: Length of FAVORITE_ARTIST
    artist_length = len(favorite_artist)
    
    return seed_digit, id_sum, name_length, artist_length


def generate_datasets(seed_digit, id_sum, artist_length, name_length, size=10):
    """Generates two distinct sorted lists (ascending) using student inputs."""
    # Ensure step sizes are at least 1 to create strictly ascending values
    step_a = max(1, id_sum % 5)
    step_b = max(1, name_length % 5)
    
    # Dataset A generation
    dataset_a = [seed_digit + (i * step_a) for i in range(size)]
    
    # Dataset B generation
    dataset_b = [artist_length + (i * step_b) for i in range(size + 2)] # slightly different size
    
    return dataset_a, dataset_b


def consolidate_records(list_a, list_b):
    """
    Consolidates two sorted lists into one ordered collection using a two-pointer approach.
    Counts the exact number of element-to-element comparisons performed.
    """
    consolidated = []
    i, j = 0, 0
    comparison_count = 0
    
    # Merge elements until one list is exhausted
    while i < len(list_a) and j < len(list_b):
        comparison_count += 1
        if list_a[i] <= list_b[j]:
            consolidated.append(list_a[i])
            i += 1
        else:
            consolidated.append(list_b[j])
            j += 1
            
    # Append any remaining elements (no comparisons needed here)
    while i < len(list_a):
        consolidated.append(list_a[i])
        i += 1
        
    while j < len(list_b):
        consolidated.append(list_b[j])
        j += 1
        
    return consolidated, comparison_count


def verify_results(list_a, list_b, consolidated):
    """Verifies the total record count and correct sorted order."""
    expected_count = len(list_a) + len(list_b)
    actual_count = len(consolidated)
    
    # Check 1: Record count verification
    count_valid = (expected_count == actual_count)
    
    # Check 2: Sorted order verification
    is_sorted = all(consolidated[i] <= consolidated[i + 1] for i in range(len(consolidated) - 1))
    
    return count_valid, is_sorted, expected_count, actual_count


# =====================================================================
# MAIN EXECUTION
# =====================================================================
if __name__ == "__main__":
    print("=== UNIVERSITY ENROLLMENT CONSOLIDATION SYSTEM ===\n")
    
    # --- STEP 0: Student Input (Replace placeholders with your real details) ---
    LAST_NAME = "Doe"
    STUDENT_ID = "202410345"
    FAVORITE_ARTIST = "DaVinci"
    
    # Calculate student parameters
    seed, id_sum, name_len, artist_len = calculate_student_inputs(LAST_NAME, STUDENT_ID, FAVORITE_ARTIST)
    
    # --- STEP 1: Generate Datasets ---
    dataset_a, dataset_b = generate_datasets(seed, id_sum, artist_len, name_len)
    
    # --- STEP 2 & 3: Consolidate and Count Comparisons ---
    consolidated_list, comparisons = consolidate_records(dataset_a, dataset_b)
    
    # --- STEP 4: Verification ---
    count_ok, order_ok, exp_cnt, act_cnt = verify_results(dataset_a, dataset_b, consolidated_list)
    
    # --- OUTPUT ASSESSMENT DATA ---
    print("--- Assessment Data ---")
    print(f"Department A Dataset: {dataset_a}")
    print(f"Department B Dataset: {dataset_b}")
    print(f"Consolidated Dataset: {consolidated_list}")
    print(f"Number of records before consolidation: {exp_cnt}")
    print(f"Number of records after consolidation:  {act_cnt}")
    print(f"Comparison Count: {comparisons}")
    
    print("\n--- Ordering Verification ---")
    print(f"✔ Record count matches: {count_ok} ({act_cnt} total records)")
    print(f"✔ Ordered sequence is verified ascending: {order_ok}")
    
    # --- STEP 5: Complexity Analysis Output ---
    print("\n--- Time and Space Complexity Analysis ---")
    print("• Time Complexity:")
    print("  - Best Case:  O(N) where N is the total elements of the smaller array (if all elements are smaller).")
    print("  - Worst Case: O(N + M) where elements alternate, requiring scanning through both input sizes completely.")
    print("• Space Complexity:")
    print("  - O(N + M) auxiliary space to construct and store the newly consolidated list.")
```

```python
import time
import random
import copy

# =====================================================================
# 1. STUDENT-SPECIFIC INPUTS (Replace these with your personal details)
# =====================================================================
LAST_NAME = "SMITH"          # Your surname
STUDENT_ID = "12345678"      # Your student ID number
FAVORITE_ARTIST = "PICASSO"  # Your favorite artist

# Automatically calculate rule-based values
SEED_DIGIT = int(STUDENT_ID[-1]) if STUDENT_ID else 0
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(char.upper()) - 64 for char in FAVORITE_ARTIST if char.isalpha())

print("--- Student-Specific Derived Values ---")
print(f"Seed Digit: {SEED_DIGIT}")
print(f"ID Sum: {ID_SUM}")
print(f"Name Length: {NAME_LENGTH}")
print(f"Artist Length: {ARTIST_LENGTH}")
print(f"Artist Sum: {ARTIST_SUM}\n")


# =====================================================================
# 2. DATA GENERATION
# =====================================================================
# Using derived student values to determine a deterministic dataset size
# and random seed for reproducibility.
DATASET_SIZE = (ID_SUM * 5) + 100  # Adjust size as needed for execution time noticeable variations
random.seed(SEED_DIGIT)

# Generate base "Orders" (e.g., tracking simulated Order IDs or values)
base_orders = [random.randint(1000, 9999) + ARTIST_SUM for _ in range(DATASET_SIZE)]

# Generate the three input conditions from the same base orders
inputs = {
    "Sorted": sorted(base_orders),
    "Reverse-ordered": sorted(base_orders, reverse=True),
    "Random": base_orders.copy()
}


# =====================================================================
# 3. SORTING ALGORITHMS WITH METRIC TRACKING
# =====================================================================

def bubble_sort(arr):
    n = len(arr)
    comparisons = 0
    movements = 0  # Counting each item assignment/shift
    
    start_time = time.perf_counter()
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            comparisons += 1
            if arr[j] > arr[j + 1]:
                # Swap elements (3 movements)
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                movements += 3
                swapped = False
        if swapped:
            break
    end_time = time.perf_counter()
    
    return comparisons, movements, (end_time - start_time)


def selection_sort(arr):
    n = len(arr)
    comparisons = 0
    movements = 0
    
    start_time = time.perf_counter()
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            comparisons += 1
            if arr[j] < arr[min_idx]:
                min_idx = j
                
        if min_idx != i:
            # Swap elements (3 movements)
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
            movements += 3
    end_time = time.perf_counter()
    
    return comparisons, movements, (end_time - start_time)


def insertion_sort(arr):
    n = len(arr)
    comparisons = 0
    movements = 0
    
    start_time = time.perf_counter()
    for i in range(1, n):
        key = arr[i]
        movements += 1  # Key assignment
        j = i - 1
        
        while j >= 0:
            comparisons += 1
            if arr[j] > key:
                arr[j + 1] = arr[j]
                movements += 1  # Shift movement
                j -= 1
            else:
                break
        else:
            # If the loop ended because j >= 0 became false, 
            # we don't have an extra failed comparison inside the else/break structure.
            pass
            
        arr[j + 1] = key
        movements += 1  # Place key back
    end_time = time.perf_counter()
    
    return comparisons, movements, (end_time - start_time)


# =====================================================================
# 4. EXECUTION AND VERIFICATION
# =====================================================================
results = []
algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

for algo_name, algo_func in algorithms.items():
    for condition, data in inputs.items():
        # Work on a deep copy to keep original conditions intact
        data_copy = copy.deepcopy(data)
        
        # Run sorting
        comps, moves, exec_time = algo_func(data_copy)
        
        # Verification requirement
        is_correctly_sorted = data_copy == sorted(base_orders)
        
        results.append({
            "Algorithm": algo_name,
            "Condition": condition,
            "Comparisons": comps,
            "Movements": moves,
            "Time (s)": f"{exec_time:.6f}",
            "Verified": "PASSED" if is_correctly_sorted else "FAILED"
        })

# =====================================================================
# 5. PERFORMANCE ASSESSMENT REPORT OUTPUT
# =====================================================================
print(f"--- Performance Report (Dataset Size: {DATASET_SIZE}) ---")
print(f"{'Algorithm':<16} | {'Condition':<16} | {'Comparisons':<12} | {'Movements':<12} | {'Time (s)':<10} | {'Verification'}")
print("-" * 80)
for r in results:
    print(f"{r['Algorithm']:<16} | {r['Condition']:<16} | {r['Comparisons']:<12} | {r['Movements']:<12} | {r['Time (s)']:<10} | {r['Verified']}")
```
